```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime

# Use seaborn style directly
sns.set(style="whitegrid")

# Get the ticker data for Microsoft
msft = yf.Ticker("HDFCBANK.NS")

# Get the dividends data
df = msft.dividends
df = df.reset_index()
df['Date'] = pd.to_datetime(df['Date'])
df.index = df['Date']


# Ensure the Dividends column is numeric
df['Dividends'] = pd.to_numeric(df['Dividends'])

# Resample the data yearly and sum the dividends, excluding the Date column
data = df[['Dividends']].resample('Y').sum()
data = data.reset_index()
df.head(12)
# Extract the year from the date
data['Year'] = data['Date'].dt.year

# Plotting
plt.figure()
plt.plot(data['Year'], data['Dividends'], 'o-')
plt.xlabel('Year')
plt.ylabel('Total Dividends (Rs.)')
plt.title('Total Dividends vs Year')
plt.grid(True)
plt.show()
```

```python
import yfinance as yf
import pandas as pd

# Get the ticker data for HDFC Bank
hdfc = yf.Ticker("HDFCBANK.NS")

# Get the expiration dates for the options
expirations = hdfc.options
print("Available expiration dates:", expirations)

# Choose a specific expiration date (for example, the first available one)
# Ensure that the date exists in the expirations list
expiration = '2024-08-09'
if expiration in expirations:
    # Get the options chain for the chosen expiration date
    options_chain = hdfc.option_chain(expiration)

    # Extract the call options data
    call_options = options_chain.calls

    # Display the call options data
    print(call_options)
else:
    print(f"The expiration date {expiration} is not available. Please choose a valid date.")
```
